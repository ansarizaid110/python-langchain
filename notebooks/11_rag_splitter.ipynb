```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
text = "Graphic designers and publishers often use sample text to test layouts and ensure that " \
    "the design works well with the content. Whether it’s a website, brochure, or book, sample text helps in assessing how readable the content will be once the final text is added."
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size = 50, chunk_overlap = 20)
```

```python
chunks = splitter.split_text(text)
```
