#### Text Document Loader

```python
from langchain_community.document_loaders import TextLoader
loader = TextLoader("../data/speech.txt")
texts = loader.load()
len(texts)
textData = texts[0]
print(textData.page_content)
```

#### PDF Document Loader

```python
from langchain_community.document_loaders import PyPDFLoader
loader = PyPDFLoader("../data/sample.pdf")
docs = loader.load()
len(docs)
oneDoc = docs[0]
print(oneDoc)
```

#### CSV File Loader

```python
from langchain_community.document_loaders import CSVLoader
loader = CSVLoader("../data/sample.csv")
docs = loader.load()
len(docs)
docs[:5]
```

#### Webpage Loader

```python
from langchain_community.document_loaders import WebBaseLoader
loader = WebBaseLoader(web_path="https://www.techsimplus.com/programs/Java-Full-Stack-Development-course-in-bhopal")
docs = loader.load()
docs[0].page_content
```
